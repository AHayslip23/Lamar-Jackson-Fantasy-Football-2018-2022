```python
#clusters between completions and attempts
```

```python
cluster1 = lamarjackson[["Completions", "Attempts"]] 
```

```python
cluster1
```

```python
X = cluster1.iloc[:, [0, 1]].values 
```

```python
X
```

```python
from sklearn.cluster import KMeans 
wcss = []
for i in range(1, 11):
    kmeans = KMeans(n_clusters = i, init="k-means++", random_state=42)
    kmeans.fit(X)
    wcss.append(kmeans.inertia_)
plt.plot(range(1, 11), wcss)
plt.title("The Elbow Method")
plt.xlabel("Number of Clusters")
plt.ylabel("WCSS")
plt.show() 
```

```python
#use three clusters 
```

```python
kmeans = KMeans(n_clusters=3, init="k-means++", random_state=42)
y_kmeans = kmeans.fit_predict(X) 
```

```python
plt.scatter(X[y_kmeans == 0, 0], X[y_kmeans == 0, 1], s=100, c="red", label="Cluster 1")
plt.scatter(X[y_kmeans == 1, 0], X[y_kmeans == 1, 1], s=100, c="blue", label="Cluster 2")
plt.scatter(X[y_kmeans == 2, 0], X[y_kmeans == 2, 1], s=100, c="purple", label="Cluster 3")
plt.scatter(kmeans.cluster_centers_[:, 0], kmeans.cluster_centers_[:, 1], s=100, c="yellow", label="Centroids")
plt.xlabel("Completions")
plt.ylabel("Attempts")
plt.legend()
plt.show() 
```

```python
#cluster 1: between 15 and 20 completions and 20 and 30 attempts
#cluster 2: between 10 to 15 completions and 15 to 25 attempts 
#cluster 3: between 20 to 25 completions and 35 to 40 attempts 
#generally he is a moderately efficent passer 
```
